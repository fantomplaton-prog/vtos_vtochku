```python
import struct

# Способ 1: Используя struct.unpack
def bytes_to_float(byte_array):
    # Little-endian (стандарт для x86, ARM)
    result_le = struct.unpack('<f', bytes(byte_array))[0]
    return result_le

# Обратная функция: float -> bytes
def float_to_bytes(value):
    # Little-endian (стандарт для x86, ARM)
    return list(struct.pack('<f', value))

print(f"Проверка обратного преобразования:")
for test_value in [-7.0, -1.0, 0.0, 1.0, 9876543210, 1234567890]:
    test_bytes = float_to_bytes(test_value)
    print(f"{test_value} -> bytes: {list(test_bytes)}")
    print(f"{list(test_bytes)} -> {bytes_to_float(list(test_bytes))}")
    assert(bytes_to_float(list(test_bytes)) == test_value)
```

```python
import struct

# The float value to convert
float_value = 123456789.123456789

# Convert float to bytes (e.g., as a 4-byte float in big-endian byte order)
# '>' specifies big-endian, 'f' specifies a single-precision float (4 bytes)
# 'd' specifies a double-precision float (8 bytes)
packed_bytes = struct.pack('>f', float_value) 
print(f"Packed bytes (4-byte float): {packed_bytes}")

packed_double_bytes = struct.pack('>d', float_value)
print(f"Packed bytes (8-byte double): {packed_double_bytes}")

# add reverse functions
def bytes_to_float(byte_array):
    """Unpack 4 bytes as big-endian float."""
    if len(byte_array) != 4:
        raise ValueError("Input must be exactly 4 bytes for float ('f').")
    return struct.unpack('>f', bytes(byte_array))[0]

def bytes_to_double(byte_array):
    """Unpack 8 bytes as big-endian double."""
    if len(byte_array) != 8:
        raise ValueError("Input must be exactly 8 bytes for double ('d').")
    return struct.unpack('>d', bytes(byte_array))[0]

print(f"Bytes to float: {bytes_to_float(packed_bytes)}")
print(f"Bytes to double: {bytes_to_double(packed_double_bytes)}")
```
