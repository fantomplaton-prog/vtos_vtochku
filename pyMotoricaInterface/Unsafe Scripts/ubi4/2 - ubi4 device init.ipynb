```python
SERIAL_TIMEOUT_MS = 100 # ms

ser = serial.Serial(
    port='/dev/cu.usbmodem1234561',
    baudrate=115200,
    timeout=SERIAL_TIMEOUT_MS/1000
)

time.sleep(1)
ser.read(ser.in_waiting)
```

```python
from ubi4_stack import Ubi4Stack
        
    
ubi4_stack = Ubi4Stack(ser)

display(ubi4_stack.request_first_info_blocked())

print('********************************')

for parameter in ubi4_stack.request_device_parameters():
    print('--------------------------------')
    print(f'Parameter Found: ID = {parameter.ID}')

print('********************************')

for sub_device in ubi4_stack.request_sub_device_info():
    print('--------------------------------')
    print('Subdevice Found:')
    display(sub_device)
    
print('********************************')

```
